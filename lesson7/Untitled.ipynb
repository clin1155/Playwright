```python
a = 6
while a > 0:
    print(a)
    a -= 1 
print("離開後 a 為", a)
```

```python
s=10
while s>1:
    print(s)
    s-=1
print(s)
```

```python
import numpy as np
for k in range(1001,0,-1):
    if np.sqrt(k)<5:
        print(k)
```

```python
odd_sum=0
even_sum=0
for i in range(1,1001):
    if i % 2 ==0:
        even_sum+=i
    else:
        odd_sum+=i
print(odd_sum)
print(even_sum)
    
```

```python
while True:
    stuff = input("請輸入小寫英文字[按q會離開]:")
    if stuff == 'q':
        break
    print(stuff.capitalize())
print("程式結束")
```

```python
total=0
while True:
    result=int(input("請輸入整數"))
    total += result
    print(total)
    if total>=500:
        break
```

```python
total:int = 0
num:int = 0
while True:
    input_value = int(input(f"請輸入第{num + 1}個數值:"))    
    if input_value < 0:
        break
        
    num += 1

    if input_value % 2 == 1:
        continue

    total += input_value

print(f"輸入的次數是{num}, ，所有輸入的正偶數的加總是:{total}")
```
